#### **Import Libraries & Configure Display**

```python
import pandas as pd
import numpy as np
import os

# Configure display options to see all columns
pd.set_option('display.max_columns', None)
pd.set_option('display.max_rows', 100)

print("Libraries loaded successfully!")
```

#### **Define File Paths & Load Datasets**

```python
RAW_DIR = "../data/raw"

ember_path = os.path.join(RAW_DIR, "ember_yearly_generation.csv")
wb_path = os.path.join(RAW_DIR, "world_bank_electricity_access.csv")
irena_path = os.path.join(RAW_DIR, "irenastat_capacity.csv")

# Load datasets
df_ember = pd.read_csv(ember_path, low_memory=False)
df_wb = pd.read_csv(wb_path, low_memory=False)
df_irena = pd.read_csv(irena_path, low_memory=False)

print("Datasets loaded successfully into memory.")
```

#### **Inspect Shape & First Few Rows of Ember Data**

```python
print(f"Ember Shape: {df_ember.shape}")
df_ember.head(3)
```

#### **Inspect Ember Data Types & Missing Values**

```python
print("--- EMBER INFO ---")
print(df_ember.info())

print("\n--- EMBER MISSING VALUES (%) ---")
missing_ember = (df_ember.isnull().mean() * 100).round(2)
print(missing_ember[missing_ember > 0])
```

#### **Inspect World Bank Electricity Access Data**

```python
print(f"World Bank Shape: {df_wb.shape}")
df_wb.head(3)
```

#### **Inspect IRENASTAT Capacity Data**

```python
print(f"IRENA Shape: {df_irena.shape}")
print(df_irena.info())
df_irena.head(3)
```

#### **Check Unique Countries & Coverage in Ember**

```python
# Check unique countries in Ember dataset
if 'Area' in df_ember.columns:
    print(f"Total unique regions in Ember: {df_ember['Area'].nunique()}")
    print(df_ember['Area'].unique()[:15]) # Previewing first 15
elif 'Country' in df_ember.columns:
    print(f"Total unique regions in Ember: {df_ember['Country'].nunique()}")
    print(df_ember['Country'].unique()[:15])
```
